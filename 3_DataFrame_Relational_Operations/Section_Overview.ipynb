# 3. DataFrame Relational Operations
### Duration - 3 Minutes

This section covers combining DataFrames: join types and join conditions, set operations, and the performance considerations - broadcast joins, data skew, and projection - that matter when joining distributed data. The demo joins the Bakehouse sample tables to answer real questions about customers, franchises, and suppliers.

## Learning Objectives

**By the end of this section, you will be able to:**

- Apply different types of DataFrame joins (inner, outer, left, right, cross) to combine datasets
- Use set operations like union, intersect, and subtract to combine or compare DataFrames
- Apply best practices for optimizing relational operations, including broadcast joins and handling data skew
