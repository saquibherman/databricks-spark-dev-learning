# 2. Grouping and Aggregating Data
### Duration - 2 Minutes

This section covers the aggregation toolkit: how `groupBy()` distributes work across the cluster, the core aggregation methods, combining multiple aggregations with `agg()`, and window functions for analytics that keep row-level detail. A demo over NYC taxi data and a lab over Bakehouse e-commerce data put the concepts to work.

## Learning Objectives

**By the end of this section, you will be able to:**

- Describe how distributed grouping transformations work and map them to SQL `GROUP BY` syntax
- Apply and combine multiple aggregation functions using both DataFrame methods and SQL expressions
- Use window functions for advanced analytics including ranking, running calculations, and comparative analysis across rows
