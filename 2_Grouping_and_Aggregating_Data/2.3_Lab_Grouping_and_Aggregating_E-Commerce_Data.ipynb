## D. Multiple Aggregations

Finance wants a payment method summary: total revenue, average transaction value, and transaction count for each payment method.

### Exercise D1: Payment Method Summary

Group the transactions by `paymentMethod` and compute, in a single `agg()` call:

- the sum of `totalPrice` rounded to 2 decimal places, aliased as `total_revenue`
- the average of `totalPrice` rounded to 2 decimal places, aliased as `avg_transaction_value`
- the count of all rows, aliased as `transaction_count`

Order the result by `total_revenue` descending and display it.

```python
# D1. Aggregate by Payment Method (Your Turn)
from pyspark.sql.functions import sum, avg, count, round, desc

payment_summary_df = (
    transactions_df
    .groupBy(FILL_IN)
    .agg(
        FILL_IN.alias("total_revenue"),
        FILL_IN.alias("avg_transaction_value"),
        FILL_IN.alias("transaction_count")
    )
    .orderBy(FILL_IN)
)

display(payment_summary_df)
```

## E. Bonus Challenge: Rank Products by Revenue

Combine grouping with a window function.

### Exercise E1: Revenue Ranking

Group the transactions by `product`, sum `totalPrice` as `product_revenue`, then use a window function to add a `revenue_rank` column ranking products from highest to lowest revenue. Display the result ordered by rank.

```python
# E1. Rank Products by Revenue (Your Turn)
from pyspark.sql.window import Window
from pyspark.sql.functions import sum, rank, desc

product_revenue_df = (
    transactions_df
    .groupBy(FILL_IN)
    .agg(FILL_IN.alias("product_revenue"))
)

revenue_window = Window.orderBy(FILL_IN)

ranked_products_df = product_revenue_df.withColumn("revenue_rank", FILL_IN)

display(ranked_products_df.orderBy("revenue_rank"))
```
