# 1. DataFrame API Basics
### Duration - 2 Minutes

This section refreshes the core DataFrame transformation toolkit you will use throughout the course: transformation methods and their SQL equivalents, column references, built-in functions, missing value handling, and user defined functions. An optional demo applies all of it in an end-to-end ETL pipeline.

## Learning Objectives

**By the end of this section, you will be able to:**

- Describe how distributed DataFrame transformations map to equivalent SQL operations
- Reference columns correctly and use Column object methods and built-in functions
- Handle missing and invalid values, including Spark 4 ANSI mode behavior
- Create and apply user defined functions, including vectorized pandas UDFs
