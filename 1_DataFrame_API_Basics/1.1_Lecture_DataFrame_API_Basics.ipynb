## A. DataFrame Transformation Methods

The core methods for distributed data transformation:

- DataFrame operations are distributed automatically across all partitions, enabling parallel processing of large datasets
- All DataFrame transformations have equivalent SQL operations under the hood (e.g. `select()` maps to `SELECT`, `filter()` maps to `WHERE`)
- Methods return a new DataFrame rather than modifying the existing one, supporting immutable transformations
- Each operation builds upon the logical plan until an action triggers execution

| DataFrame Method | SQL Equivalent | Description |
|------------------|----------------|-------------|
| __`select()`__ | `SELECT` | Choose specific columns |
| __`filter()`__, `where()` | `WHERE` | Filter rows based on a condition |
| __`groupBy()`__ | `GROUP BY` | Group rows for aggregation |
| __`orderBy()`__, `sort()` | `ORDER BY` | Sort a DataFrame by columns |
| __`join()`__ | `JOIN` | Combine two DataFrames by a key |

For more information see the [DataFrame API reference](https://spark.apache.org/docs/latest/api/python/reference/pyspark.sql/dataframe.html).

## B. Referencing DataFrame Columns

There are several ways to reference a column, and they are not all equal:

| Type | Description | Example |
|------|-------------|---------|
| __Direct__ | Simplest syntax, works for basic column selection | `df.select("name")` |
| __By attribute__ | Only works for column names that are valid Python identifiers | `df.select(df.first_name)` |
| __Column expression__ | Works with any column name, can be referenced across DataFrames (e.g. in joins) | `df.select(df["first-name"])` |
| __Column object__ | Required when building complex expressions or using column-specific operations like `alias()`, `cast()`, `asc()` or `desc()` | `df.select(col("name").alias("customer_name"))` |

### Common Column Object Methods

| Method | Description | Example |
|--------|-------------|---------|
| __`alias()`__ | Rename a column | `col("name").alias("customer_name")` |
| __`cast()`__ / __`astype()`__ | Change data type | `col("age").cast("int")` |
| __`isNull()`__ / __`isNotNull()`__ | Check for nulls | `col("name").isNotNull()` |
| __`contains()`__ | String matching | `col("title").contains("Manager")` |
| __`asc()`__ / __`desc()`__ | Sort direction (used with `sort` / `orderBy`) | `df.sort(col("c1").desc(), col("c2").asc())` |

For more information see the [Column API reference](https://spark.apache.org/docs/latest/api/python/reference/pyspark.sql/column.html).

## C. Built-in Functions

Spark ships with a large library of functions that operate on columns:  

- Functions can operate on scalar (strings, ints, etc.) or complex data types (arrays, structs, etc.)
- Function categories include Math, Datetime, String, Collection, Aggregate, Window, and more
- The same functions are available in both DataFrame operations and SQL queries using the equivalent SQL function names

| Built-in Function | SQL Equivalent | Description |
|-------------------|----------------|-------------|
| __`round(col, scale)`__ | `ROUND` | Round a number to a scale |
| __`concat(col1, col2)`__ | `CONCAT` | Concatenate strings |
| __`date_format(col, fmt)`__ | `DATE_FORMAT` | Format dates as strings |
| __`regexp_replace(col, pattern, replace)`__ | `REGEXP_REPLACE` | Replace using a regex |
| __`coalesce(col1, col2)`__ | `COALESCE` | First non-null value |

For more information see the [functions reference](https://spark.apache.org/docs/latest/api/python/reference/pyspark.sql/functions.html) and the [SQL function reference](https://spark.apache.org/docs/latest/api/sql/index.html).

## D. Handling Missing Values

<p style="font-size: 1em; line-height: 1.6; color: #333">Real data has gaps. Common functions for handling nulls:</p>

- `isNull()` / `isNotNull()` - check if values are null
- `count(col)` - counts non-null values in a specific column
- `df.fillna()` / `df.na.fill()` - replace nulls with values
- `df.dropna()` / `df.na.drop()` - remove rows with nulls

<div class="code-block" data-language="python">
# Drop rows with nulls in a specific column
df.dropna(subset=["my_column"])
<br/>
# ..or..
df.na.drop(subset=["my_column"])
<br/>
# Replace nulls with a value
df.fillna("unknown")
<br/>
# ..or..
df.na.fill("unknown")
</div>

## E. Type Conversion Under ANSI Mode

Spark 4 enables <strong>ANSI SQL mode</strong> by default, which changes how invalid conversions behave:

| Expression | Spark 3 (ANSI off) | Spark 4 (ANSI on) |
|------------|--------------------|--------------------|
| __`CAST('NA' AS INT)`__ | Returns `NULL` silently | Raises an error |
| __`try_cast('NA' AS INT)`__ | Returns `NULL` | Returns `NULL` |
| Integer overflow | Wraps around silently | Raises an error |

## F. User Defined Functions

<p style="font-size: 1em; line-height: 1.6; color: #333">UDFs allow you to apply your own Python functions to DataFrame columns:</p>

<div class="code-block" data-language="python">
from pyspark.sql.functions import udf
<br/>
# define your function
@udf("string")
def make_greeting(name):
    return f"Hello, {name}!"
<br/>
# use your function
df.select(make_greeting("name"))
</div>

- UDFs allow developers to create reusable custom functions
- UDFs can adversely impact performance: they cannot be optimized by the Catalyst optimizer and add serialization overhead between the JVM and the Python worker
- Always use a built-in function if one is available

### Pandas UDFs and Apache Arrow

<p style="font-size: 1em; line-height: 1.6; color: #333">Pandas UDFs let you write Python functions that operate on <strong>batches of rows</strong> instead of single rows, leveraging Apache Arrow for more efficient Python-JVM serialization.</p>

<div class="code-block" data-language="python">
from pyspark.sql.functions import pandas_udf
<br/>
# define your pandas udf
@pandas_udf("integer")
def age_plus_one(age_series):
    return age_series + 1
<br/>
# use your pandas udf
df.select(age_plus_one("age"))
</div>

- More efficient than their non-pandas counterparts
- In Spark 4 you can also accelerate row-based UDFs with Arrow serialization: `@udf("string", useArrow=True)`
- You should still use a built-in function if one is available - built-ins are always fastest because the optimizer understands them

## Summary

- DataFrame transformations map one-to-one to SQL operations and are distributed across partitions automatically
- Use Column objects (`col()`) for complex expressions and column methods like `alias()` and `cast()`
- Prefer built-in functions; reach for pandas UDFs when custom logic is unavoidable
- Spark 4 ANSI mode makes invalid casts fail loudly; `try_cast()` provides the permissive alternative for cleaning dirty data

            html += '<button id="retakeBtn" style="padding: 14px 32px; background: #1976d2; color: white; border: none; border-radius: 8px; cursor: pointer; font-size: 1.1em; font-weight: 600;">Retake Quiz</button>';
            html += '</div>';
        } else {
            var q = questions[currentIndex];
            
            // Question
            html += '<div style="background: #fff; border-radius: 12px; border: 2px solid #e0e0e0; padding: 24px; margin-bottom: 20px;">';
            html += '<div style="font-size: 1.15em; color: #333; line-height: 1.6;">' + q.question + '</div>';
            html += '</div>';
            
            // Options
            html += '<div style="display: flex; flex-direction: column; gap: 12px; margin-bottom: 20px;">';
            
            q.options.forEach(function(opt) {
                var isSelected = selectedAnswer === opt.id;
                var isCorrect = opt.id === q.correct;
                var bgColor = '#fff';
                var borderColor = '#e0e0e0';
                var textColor = '#333';
                
                if (showingFeedback) {
                    if (isCorrect) {
                        bgColor = '#e8f5e9';
                        borderColor = '#4caf50';
                        textColor = '#2e7d32';
                    } else if (isSelected && !isCorrect) {
                        bgColor = '#ffebee';
                        borderColor = '#f44336';
                        textColor = '#c62828';
                    }
                } else if (isSelected) {
                    bgColor = '#e3f2fd';
                    borderColor = '#1976d2';
                    textColor = '#1565c0';
                }
                
                var cursor = showingFeedback ? 'default' : 'pointer';
                
                html += '<div class="mcq-option" data-id="' + opt.id + '" style="display: flex; align-items: center; gap: 16px; padding: 16px 20px; background: ' + bgColor + '; border: 2px solid ' + borderColor + '; border-radius: 8px; cursor: ' + cursor + '; transition: all 0.15s ease;">';
                html += '<div style="width: 32px; height: 32px; border: 2px solid ' + borderColor + '; border-radius: 50%; display: flex; align-items: center; justify-content: center; font-weight: 700; color: ' + textColor + '; background: ' + (isSelected || (showingFeedback && isCorrect) ? borderColor : 'transparent') + '; color: ' + (isSelected || (showingFeedback && isCorrect) ? '#fff' : textColor) + ';">' + opt.id.toUpperCase() + '</div>';
                html += '<div style="flex: 1; font-size: 1em; color: ' + textColor + ';">' + opt.text + '</div>';
                if (showingFeedback && isCorrect) {
                    html += '<span style="color: #4caf50; font-size: 1.2em;">✓</span>';
                } else if (showingFeedback && isSelected && !isCorrect) {
                    html += '<span style="color: #f44336; font-size: 1.2em;">✗</span>';
                }
                html += '</div>';
            });
            
            html += '</div>';
            
            // Feedback
            if (showingFeedback) {
                var isCorrect = selectedAnswer === q.correct;
                html += '<div style="background: ' + (isCorrect ? '#e8f5e9' : '#fff3e0') + '; border: 2px solid ' + (isCorrect ? '#4caf50' : '#ff9800') + '; border-radius: 8px; padding: 16px 20px; margin-bottom: 20px;">';
                html += '<div style="font-weight: 600; color: ' + (isCorrect ? '#2e7d32' : '#e65100') + '; margin-bottom: 8px;">' + (isCorrect ? '✓ Correct!' : '✗ Not quite') + '</div>';
                html += '<div style="color: #333; font-size: 0.95em; line-height: 1.5;">' + q.explanation + '</div>';
                html += '</div>';
            }
            
            // Buttons
            html += '<div style="display: flex; gap: 12px; justify-content: flex-end;">';
            if (!showingFeedback) {
                html += '<button id="submitBtn" style="padding: 12px 28px; background: ' + (selectedAnswer ? '#4caf50' : '#ccc') + '; color: white; border: none; border-radius: 6px; cursor: ' + (selectedAnswer ? 'pointer' : 'not-allowed') + '; font-size: 1em; font-weight: 600;"' + (selectedAnswer ? '' : ' disabled') + '>Submit Answer</button>';
            } else {
                html += '<button id="nextQBtn" style="padding: 12px 28px; background: #1976d2; color: white; border: none; border-radius: 6px; cursor: pointer; font-size: 1em; font-weight: 600;">' + (currentIndex < questions.length - 1 ? 'Next Question' : 'See Results') + '</button>';
            }
            html += '</div>';
        }
        
        html += '</div>';
        
        root.innerHTML = html;
        attachEvents();
    }
    
    function attachEvents() {
        var options = document.querySelectorAll('.mcq-option');
        var submitBtn = document.getElementById('submitBtn');
        var nextQBtn = document.getElementById('nextQBtn');
        var retakeBtn = document.getElementById('retakeBtn');
        
        options.forEach(function(opt) {
            opt.addEventListener('click', function() {
                if (!showingFeedback) {
                    selectedAnswer = this.dataset.id;
                    render();
                }
            });
            
            if (!showingFeedback) {
                opt.addEventListener('mouseenter', function() {
                    if (selectedAnswer !== this.dataset.id) {
                        this.style.borderColor = '#1976d2';
                        this.style.background = '#f5f9ff';
                    }
                });
                opt.addEventListener('mouseleave', function() {
                    if (selectedAnswer !== this.dataset.id) {
                        this.style.borderColor = '#e0e0e0';
                        this.style.background = '#fff';
                    }
                });
            }
        });
        
        if (submitBtn) {
            submitBtn.addEventListener('click', function() {
                if (selectedAnswer) {
                    var q = questions[currentIndex];
                    if (selectedAnswer === q.correct) {
                        score++;
                    }
                    showingFeedback = true;
                    render();
                }
            });
        }
        
        if (nextQBtn) {
            nextQBtn.addEventListener('click', function() {
                currentIndex++;
                selectedAnswer = null;
                showingFeedback = false;
                render();
            });
        }
        
        if (retakeBtn) {
            retakeBtn.addEventListener('click', function() {
                init();
            });
        }
    }
    
    if (typeof NodeList.prototype.forEach !== 'function') {
        NodeList.prototype.forEach = Array.prototype.forEach;
    }
    
    init();
})();
