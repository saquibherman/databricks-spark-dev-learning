# 4. Working with Complex Data
### Duration - 2 Minutes

This section covers Spark's complex data types - arrays, structs, and maps - and the functions that operate on them: converting JSON strings to structured types, accessing struct fields, unnesting arrays with `explode()`, aggregating values into collections, and pivoting. The demo and lab work with JSON-string tables built from the Bakehouse sample data during the required setup.

## Learning Objectives

**By the end of this section, you will be able to:**

- Describe Spark's support for arrays, maps, and structs, and where these types appear in real-world data
- Convert JSON strings to structured types with `schema_of_json()` and `from_json()`
- Manipulate array columns using functions like `explode()`, `array_contains()`, and `element_at()`
- Aggregate values into collections with `collect_list()` and `collect_set()`
- Work with nested struct types, including accessing fields and flattening structures
