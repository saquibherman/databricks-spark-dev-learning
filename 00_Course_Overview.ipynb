# Developing Applications with Apache Spark
### Duration - 2 Minutes

This course builds on introductory Spark knowledge to develop practical data processing applications with the Apache Spark 4 DataFrame API on the Databricks Data Intelligence Platform. You will group and aggregate large datasets, combine DataFrames with joins and set operations, and work with complex data types including arrays, structs, and JSON. Hands-on demonstrations and labs reinforce each topic using the Databricks sample datasets.

## Terminal Learning Objective

**Develop data processing applications using the Apache Spark 4 DataFrame API on Databricks**

By the end of this course, you will be able to:

- Apply DataFrame transformation methods, column references, and built-in functions, and map them to their SQL equivalents
- Group, aggregate, and analyze large datasets, including window functions for row-level analytics
- Combine DataFrames using joins and set operations, and apply join performance best practices
- Operate on complex data types - arrays, structs, and maps - and convert JSON strings to structured types
- Apply user defined functions, including vectorized pandas UDFs, where built-in functions are not available

## Prerequisites

You should meet the following prerequisites before starting this course:

- Basic programming knowledge
- Familiarity with Python
- Basic understanding of SQL queries (`SELECT`, `JOIN`, `GROUP BY`)
- Familiarity with data processing concepts
- Completion of the *Introduction to Apache Spark* course, or prior Spark or Databricks experience
